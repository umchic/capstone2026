## **Setting up perspectives and prompts**

This notebook contains the predefined human perspectives, the city variable, and the standardized prompt template for our Capstone Project.

You can use it to generate prompts for different perspectives and adapt the city variable to your selected study area.

Please keep the perspectives and prompt structure unchanged to ensure consistency and comparability across the project.

```python
# 1. Define the human perspectives

perspectives = [
    "single person living in",
    "member of a family of four with two children living in",
    "domestic student studying in",
    "international student studying in",
    "child (6-12 years old) living in",
    "young adult (18-30 years old) living in",
    "older adult (65+ years old) living in",
    "first-time tourist visiting",
    "recent immigrant (less than 5 years) living in",
    "long-term local resident living in",
    "low-income resident living in",
    "middle-income resident living in",
    "high-income resident living in",
    "Christian resident living in",
    "Muslim resident living in",
    "Hindu resident living in"
]
```

```python
# 2. Define the city and prompt template
city_name = "Washington DC, USA"

prompt_template = """
Imagine you are a {perspective} {city_name}.
Describe the city from your own perspective, based on your
lifestyle, daily routines, and personal preferences.
Identify the places, neighborhoods, streets, and landmarks
that you most strongly associate with the city or regularly
visit. Describe the activities you typically engage in,
whether during your stay or as part of your everyday life
and leisure time. Provide a comprehensive, coherent
description in paragraph form rather than as a bullet-point list.
"""
```

```python
# 3. Generate the prompts

for perspective in perspectives:
    prompt = prompt_template.format(
        perspective=perspective,
        city_name=city_name
    )

    print(prompt)
    print("\n" + "=" * 60 + "\n")
```
